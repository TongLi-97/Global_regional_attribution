```python
path = './saved_data/'

def load_xr_pickle(filename):
    with open(filename, "rb") as f:
        return pickle.load(f)

obs_all = load_xr_pickle(path + '0.1.HadCRUT5.Tas.anomalies.46AR6regions_7cont_1glob_175years_1850-2024.pkl')

obs_200runs_all = load_xr_pickle(path + '0.2.HadCRUT5.200runs.nonmasked.Tas.anomalies.46AR6regions_7cont_1glob_175years_1850-2024.pkl')

mod_all = load_xr_pickle(path + '0.4.Smoothed_His-ALL.25mods.mean.nonmasked.Tas.anomalies.46AR6regions_7cont_1glob_171years_1850-2020.pkl')

ln_mod_all = load_xr_pickle(path + '0.3.Large_ensembles.320runs.nonmasked.Tas.anomalies.46AR6regions_7cont_1glob_251years_1850-2100.pkl')

mod_45_pseudo = load_xr_pickle(path + '0.5.pseudo-model.15mod.run1-3.nonmasked.Tas.anomalies.46AR6regions_7cont_1glob_176years_1850-2025.pkl')

print_xarray_info(obs_all, obs_200runs_all, mod_all, ln_mod_all, mod_45_pseudo)
```

```python

mod_25_warm = change_ref(mod_all, 1850, 1900).sel(year = slice(2016, 2025)).mean('year').squeeze()
pseudo_15_warm = change_ref(mod_45_pseudo, 1850, 1900).sel(year = slice(2016, 2025)).mean('year').squeeze()
```

```python
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
import numpy as np

mpl.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": ["Arial", "Helvetica", "DejaVu Sans"],
    "pdf.fonttype": 42,
    "font.size": 7,
    "axes.linewidth": 0.6,
    "legend.frameon": False,
})

# Use one value per model in both groups. The three pseudo-observation runs
# from each model are averaged before plotting, giving n=15 model means.
pseudo_15_model_warm = pseudo_15_warm.groupby("model_name").mean("model_run")
assert pseudo_15_model_warm.sizes["model_name"] == 15
assert mod_25_warm.sizes["model_name"] == 25

regions = mod_25_warm.region.values
n_regions = len(regions)
data1 = [np.asarray(mod_25_warm.isel(region=i).values) for i in range(n_regions)]
data2 = [np.asarray(pseudo_15_model_warm.isel(region=i).values) for i in range(n_regions)]

FIG_S2_WIDTH_MM, FIG_S2_HEIGHT_MM = 180, 72
fig, ax = plt.subplots(
    figsize=(FIG_S2_WIDTH_MM / 25.4, FIG_S2_HEIGHT_MM / 25.4)
)
positions1 = np.arange(n_regions) - 0.18
positions2 = np.arange(n_regions) + 0.18

common = dict(
    widths=0.28,
    patch_artist=True,
    showfliers=False,  # every value is shown by the overlaid points below
    whiskerprops=dict(linewidth=0.5),
    capprops=dict(linewidth=0.5),
)
ax.boxplot(
    data1,
    positions=positions1,
    boxprops=dict(facecolor="#e3d5ca", edgecolor="#7c6a0a", linewidth=0.5),
    medianprops=dict(color="#7c6a0a", linewidth=0.8),
    **common,
)
ax.boxplot(
    data2,
    positions=positions2,
    boxprops=dict(facecolor="#ae2012", edgecolor="#7b140d", linewidth=0.5),
    medianprops=dict(color="#e5e5e5", linewidth=0.8),
    **common,
)

# Overlay every independent model mean using reproducible horizontal jitter.
rng = np.random.default_rng(42)
for i, (full_values, pseudo_values) in enumerate(zip(data1, data2)):
    ax.scatter(
        positions1[i] + rng.uniform(-0.075, 0.075, full_values.size),
        full_values,
        s=3.2, facecolor="#e3d5ca", edgecolor="#7c6a0a",
        linewidth=0.2, alpha=0.72, zorder=3,
    )
    ax.scatter(
        positions2[i] + rng.uniform(-0.075, 0.075, pseudo_values.size),
        pseudo_values,
        s=3.2, facecolor="#ae2012", edgecolor="#7b140d",
        linewidth=0.2, alpha=0.72, zorder=3,
    )

ax.set_xticks(np.arange(n_regions))
ax.set_xticklabels(mod_25_warm.abbrevs.values, rotation=90, fontsize=5)
ax.set_ylabel("2016–2025 warming relative to 1850–1900 (°C)", fontsize=7)
ax.tick_params(axis="y", labelsize=6, width=0.6, length=2.5)
ax.tick_params(axis="x", width=0.5, length=2)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

legend_handles = [
    Patch(
        facecolor="#e3d5ca", edgecolor="#7c6a0a",
        label="Full ensemble (n=25 model means)",
    ),
    Patch(
        facecolor="#ae2012", edgecolor="#7b140d",
        label="Pseudo-observations (n=15 model means)",
    ),
]
ax.legend(handles=legend_handles, fontsize=6, loc="upper left", ncol=2)
fig.tight_layout(pad=0.35)
fig.savefig(
    "./saved_figures/Fig_S2.pdf",
    format="pdf", bbox_inches="tight", pad_inches=0.02,
    metadata={"Title": "Supplementary Figure 2"},
)
plt.show()
```
