```python
def find_yr(data, threshold):
    valid = data.where(data >= threshold, drop=True)
    if valid.size == 0:
        return np.nan
    return int(valid.year.values[0])

def summarize_threshold(post_mean, post_q, threshold):
    """
    post_mean : DataArray (median / best estimate)
    post_q    : DataArray (quantiles, dim='quantile')
    """

    yr_median = find_yr(post_mean, threshold)
    yr_low    = find_yr(post_q.sel(quantile='5th'), threshold)
    yr_high   = find_yr(post_q.sel(quantile='95th'), threshold)

    result = {
        "threshold": threshold,
        "median": yr_median,
        "low": yr_low,
        "high": yr_high
    }

    print(result)

    return result

post_time_15 = summarize_threshold(post_gsat_ref18, post_warm_5_95, 1.5)
post_time_2  = summarize_threshold(post_gsat_ref18, post_warm_5_95, 2)

prior_time_15 = summarize_threshold(prior_gsat_ref18, prior_warm_5_95, 1.5)
prior_time_2  = summarize_threshold(prior_gsat_ref18, prior_warm_5_95, 2)


```

```python
ED_FIGURE_WIDTH_MM = 180
ED_FIGURE_HEIGHT_MM = 140
ED_PANEL_LABEL_SIZE = 8.0

def plot_panel(ax, data, title, colors, continent_indices, labels, y_label,
               ylim_low, ylim_high):
    mean = data.sel(quantile='mean')
    err_lower = mean - data.sel(quantile='5th')
    err_upper = data.sel(quantile='95th') - mean

    bar_width = 0.62
    group_gap = 0.8
    x_ticks, x_labels = [], []
    current_x = 0.0

    for continent, indices in continent_indices.items():
        if continent == 'Global land':
            current_x += 1.0
        start = current_x
        for idx in indices:
            colour = colors[continent]
            ax.bar(current_x, float(mean[idx]), width=bar_width,
                   color=colour, edgecolor='none', alpha=1.0, zorder=2)
            ax.errorbar(
                current_x, float(mean[idx]),
                yerr=[[float(err_lower[idx])], [float(err_upper[idx])]],
                fmt='o', ms=2.8, mfc=colour, mec='white', mew=0.4,
                ecolor='#463f3a', elinewidth=0.9, capsize=1.8,
                capthick=0.9, zorder=4,
            )
            x_ticks.append(current_x)
            x_labels.append(labels[idx])
            current_x += 1.0

        end = current_x - 1.0
        ax.axvspan(start - 0.48, end + 0.48, color='#edede9', alpha=0.45, zorder=0)
        compact_groups = {'Australasia', 'Antarctica', 'Global land'}
        group_fontsize = 5.0 if continent in compact_groups else 6.0
        ax.text((start + end) / 2, 1.025, continent,
                transform=ax.get_xaxis_transform(), ha='center', va='bottom',
                fontsize=group_fontsize, fontweight='bold', clip_on=False)
        current_x += group_gap

    ax.set_xticks(x_ticks, x_labels, rotation=90)
    ax.set_xlim(-0.7, current_x - group_gap + 1.0)
    ax.set_ylim(ylim_low, ylim_high)
    ax.set_ylabel(y_label, fontsize=7.0)
    ax.tick_params(axis='x', labelsize=5.5, length=0, pad=1.5)
    ax.tick_params(axis='y', labelsize=6.5)
    ax.grid(axis='y', color='#D9D9D9', linewidth=0.5, zorder=0)
    ax.spines[['top', 'right']].set_visible(False)
    ax.text(-0.055, 1.03, title, transform=ax.transAxes,
            fontsize=ED_PANEL_LABEL_SIZE, fontweight='bold', va='top', ha='left')


fig, axs = plt.subplots(
    2, 1,
    figsize=(ED_FIGURE_WIDTH_MM / 25.4, ED_FIGURE_HEIGHT_MM / 25.4),
    sharey=True,
)

plot_panel(
    axs[0], post_warming_15_ref2025, 'a', colors, continent_indices, labels,
    'Additional allowable warming (°C)', ylim_low=-1.5, ylim_high=3.2,
)
plot_panel(
    axs[1], post_warming_2_ref2025, 'b', colors, continent_indices, labels,
    'Additional allowable warming (°C)', ylim_low=-1.5, ylim_high=3.2,
)

fig.subplots_adjust(left=0.085, right=0.99, top=0.92, bottom=0.10, hspace=0.4)
plt.show()
fig.savefig('./saved_figures/ED_Fig6.pdf', format='pdf', metadata={'Title': 'Extended Data Figure 6'})
```

```python
def plot_grouped_bars(ax, labels, data_dict, ylabel, ylim=None, xlim=None,
                      legend_loc="upper right", title=None, legend=True):

    # Two x positions: one for 1.5°C, one for 2°C
    x = np.array([0, 1])
    width = 0.30

    # Colors
    c15 = '#bc4749'
    c2  = '#e36414'

    # Group keys by temperature target
    keys_15 = [k for k in data_dict if "1.5°C" in k]
    keys_2  = [k for k in data_dict if "2°C"  in k]
    print(keys_2)

    # Offsets for showing 2 bars in each group
    offsets = [-0.15, 0.15]   # adjust as needed

    # Plot 1.5°C bars at x=0
    for k, off in zip(keys_15, offsets):
        ax.bar(0 + off, data_dict[k], width,
               color=c15, alpha=0.7 if "Weighted" in k else 0.4,
               label=k)

    # Plot 2°C bars at x=1
    for k, off in zip(keys_2, offsets):
        ax.bar(1 + off, data_dict[k], width,
               color=c2, alpha=0.7 if "Weighted" in k else 0.4,
               label=k)

    # X-axis labels
    ax.set_xticks([0, 1])
    ax.set_xticklabels(["1.5°C", "2°C"], fontsize=8)
    ax.tick_params(axis='y', labelsize=8.0)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.grid(axis='y', alpha=0.3)



    ax.set_ylabel(ylabel, fontsize=9.0)
    if ylim: ax.set_ylim(ylim)
    if xlim: ax.set_xlim(xlim)
    if legend:
        ax.legend(loc=legend_loc, fontsize=8, frameon=False)

    if title:
        ax.set_title(title, weight='bold', loc='left', fontsize=8.0, pad=6)


# ============================================================
# MASTER FUNCTION — (1 top) + (2 bottom)
# ============================================================
def plot_global_plus_two_panels(global_constrained, global_weighted, forcings,
                                percent_data, warming_data):

    # Create layout: 2 rows, 2 columns

    # GridSpec: row heights (top tall, bottom shorter)
    gs = fig.add_gridspec(2, 2, height_ratios=[1.2, 1])

    # ---------------------------------------------------------
    # PANEL A (top, spanning 2 columns)
    # ---------------------------------------------------------
    axA = fig.add_subplot(gs[0, :])

    x = np.arange(len(forcings))
    width = 0.35

    axA.bar(x - width/2, global_weighted, width,
            label='Weighted 2016-2025',
            color='#457b9d', alpha=0.7)

    axA.bar(x + width/2, global_constrained, width,
            label='Constrained 2016-2025',
            color='#457b9d', alpha=0.4)

    axA.set_xticks(x)
    axA.set_xticklabels(forcings, rotation=0, ha='center', fontsize=8)
    axA.set_ylabel("Global land warming (°C)", fontsize=9.0)
    axA.grid(axis='y', alpha=0.3)
    axA.legend(loc="upper right", fontsize=8, frameon=False)
    axA.set_ylim(-0.6, 4.0)
    axA.spines['top'].set_visible(False)
    axA.spines['right'].set_visible(False)
    axA.set_title("a", weight="bold", fontsize=8.0, loc='left')

    # ---------------------------------------------------------
    # PANEL B — bottom left
    # ---------------------------------------------------------
    labels = ['Global land']

    axB = fig.add_subplot(gs[1, 0])
    plot_grouped_bars(
        axB, labels, percent_data,
        ylabel="Percent occurred (%)",
        ylim=(40, 100),
        xlim=(-1, 2),
        legend_loc="upper right",
        title="b"
    )

    # ---------------------------------------------------------
    # PANEL C — bottom right
    # ---------------------------------------------------------
    axC = fig.add_subplot(gs[1, 1])
    plot_grouped_bars(
        axC, labels, warming_data,
        ylabel="Attributable warming (°C)",
        ylim=(0, 4),
        xlim=(-1, 2),
        legend_loc="upper right",
        title="c"
    )

    plt.subplots_adjust(
    wspace=0.3,   # width space between columns
    hspace=0.2,   # height space between rows
    left=0.085,
    right=0.985,
    top=0.94,
    bottom=0.10)

    plt.show()
```
